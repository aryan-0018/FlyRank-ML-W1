## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research Question

**Question:** Which content pages should be prioritized for human review based on observable search, content, and engagement signals?

**Decision:** The work aims to improve the decision of which pages should be reviewed first when review capacity is limited.

**Output:** The intended output is a ranked list of content pages, supported by scores and interpretable reason codes that explain why each page was prioritized.

**Action:** A content reviewer can inspect the highest-ranked pages and decide whether each page should be refreshed, expanded, protected, monitored, or considered for another appropriate content action.

**Cost of a wrong recommendation:** A false positive could waste limited review time by prioritizing a page that does not require attention. A false negative could cause a potentially important page to be overlooked. Because of these risks, the ranking should support human decision-making rather than automatically trigger content changes.

**Why data or machine learning can help:** Page performance depends on multiple signals such as visibility, freshness, position, CTR, and engagement. A transparent manual rule provides a useful baseline, while data analysis and machine learning can test whether combining these observable signals produces a more useful prioritization. The goal is not simply to train a model, but to determine whether an evidence-based ranking can improve a real review decision and remain understandable to the person using it.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import subprocess
import pandas as pd

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if not os.path.isdir(REPO_DIR):
    subprocess.run(
        ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
        check=True
    )

data_path = os.path.join(
    REPO_DIR,
    "data",
    "raw",
    "content_refresh_anonymized.csv"
)

df = pd.read_csv(data_path)

print("Dataset loaded successfully.")
print("Rows:", len(df))
print("Columns:", len(df.columns))

declining_pages = (
    df["trend_direction"]
    .astype(str)
    .str.lower()
    .eq("down")
    .sum()
)

stale_visible_pages = (
    (df["days_since_last_update"] >= 180) &
    (df["impressions_90d"] >= 500)
).sum()

print("Total pages:", len(df))
print("Pages with a declining trend:", declining_pages)
print("Stale and visible pages:", stale_visible_pages)
```

### Quick Look at the Data

The starter dataset contains **30,000 pages**. Among them, **16,262 pages (approximately 54.2%)** have an observed `trend_direction` of "down".

Using a simple "stale and visible" rule — pages that have not been updated for at least 180 days and have at least 500 impressions over the 90-day measurement window — I found only **17 pages** that meet both conditions.

These observations make the prioritization problem interesting. While a large share of pages show a declining trend, only a small number satisfy this particular hand-written rule. This suggests that relying on a single strict rule may overlook other potentially relevant review candidates. Over the next several weeks, I want to investigate whether combining multiple observable signals can produce a more useful and explainable ranking of pages for human review.

These findings are exploratory and directional. A declining trend does not necessarily mean that a page should be refreshed, and meeting or not meeting the "stale and visible" rule does not establish that an intervention would improve future performance.

## 4. Careful words: what I can and can't claim

### What I Can and Cannot Claim

This project can identify **observed associations and directional patterns** in the provided search, content, and engagement data. It can evaluate whether these observable signals help prioritize pages for review and can provide decision-support through an explainable ranked list.

The project cannot claim that any signal is a causal Google ranking factor, nor can it claim to predict Google's ranking algorithm. A page being ranked highly for review does not prove that it needs to be changed, and a declining trend does not prove that refreshing the page will cause it to recover.

Any recommendations produced by the project should therefore be treated as **decision-support for human review**, not as automated conclusions or guarantees of future performance.
