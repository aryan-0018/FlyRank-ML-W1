## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule is an important baseline rather than something to dismiss. In the starter experiment, the hand rule achieved Precision@20 of 0.900 compared with 0.700 for the depth-2 tree, while at Precision@50 the tree achieved 0.720 compared with 0.680 for the hand rule. This suggests that a simple rule can already be useful near the very top of the queue, while a learned approach may capture additional signal further down the ranking.

ML therefore earns its place only if it can combine multiple observable signals in a way that provides a reliable improvement over a transparent baseline under honest validation. The aim is not to use ML simply because it is more sophisticated; it is to test whether the additional complexity produces a better and more useful ranking for the content-review decision.
